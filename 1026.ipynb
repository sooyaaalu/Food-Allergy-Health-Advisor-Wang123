```python
def parse_user_constraints(user_text):
    """解析用户的自然语言，输出结构化画像"""
    allergen_dict = {
        "花生": ["花生"], "牛奶": ["牛奶", "乳糖", "乳制品"],
        "鸡蛋": ["鸡蛋", "蛋清", "蛋黄"], "大豆": ["大豆", "黄豆"],
        "小麦": ["小麦", "麸质", "面筋"], "坚果": ["坚果", "杏仁", "腰果", "核桃"],
        "海鲜": ["海鲜", "虾", "蟹", "贝", "鱼"], "芝麻": ["芝麻"],
    }
    condition_dict = {
        "高血压": ["高血压", "血压高"], "高血糖": ["高血糖", "血糖高", "糖尿病"],
        "高血脂": ["高血脂", "血脂高"], "痛风": ["痛风", "尿酸高"], "肾病": ["肾病", "肾脏"],
    }
    diet_dict = {
        "低盐": ["低盐", "少盐", "清淡", "不能吃咸"],
        "低糖": ["低糖", "少糖", "控糖", "无糖", "不能吃太甜"],
        "低脂": ["低脂", "少油", "减脂"], "素食": ["素食", "吃素"],
    }
    
    allergens = [k for k, words in allergen_dict.items() if any(w in user_text for w in words)]
    conditions = [k for k, words in condition_dict.items() if any(w in user_text for w in words)]
    diet = [k for k, words in diet_dict.items() if any(w in user_text for w in words)]
    
    return {
        "allergens": allergens, "conditions": conditions,
        "diet": diet, "raw": user_text # raw 保留原话，给大模型兜底
    }

# 测试函数1
print("测试函数1：")
print(parse_user_constraints("我对花生和牛奶过敏，有高血压，不能吃太甜"))
```

```python
import re

def normalize_ingredients(raw_text):
    """清洗配料表文本，输出干净的配料列表"""
    text = re.sub(r'[（(][^）)]*[）)]', '', raw_text)       # 去括号
    text = re.sub(r'\d+(\.\d+)?\s*%', '', text)           # 去百分比
    text = re.sub(r'\b\d+[\.、\)]\s*', '', text)          # 去编号
    text = re.sub(r'^(配料表|配料|成分)[:：]?', '', text)   # 去标题
    
    parts = re.split(r'[,，、;；\n]+', text)
    result = []
    for p in parts:
        p = p.strip()
        if p and p not in result:
            result.append(p)
    return result

# 测试函数2
print("\n测试函数2：")
print(normalize_ingredients("配料表：小麦粉（≥30%）、1. 花生、食盐、白砂糖、乳粉"))
```

```python
def analyze_food(raw_ingredients, user_text):
    # ===== 3次函数调用 =====
    profile = parse_user_constraints(user_text)                # 函数1
    ingredients = normalize_ingredients(raw_ingredients)       # 函数2
    match_result = check_compatibility(ingredients, profile)   # 函数3
    
    # ===== 拼装 Prompt =====
    prompt = f"""
配料表：{ingredients}
用户原始描述：{profile['raw']}
用户结构化画像：{profile}
规则匹配结果：{match_result}

请判断这个食物是否适合该用户食用。
要求：
1. 规则匹配结果只是辅助，可能不完整。
2. 如果用户提到了规则没覆盖的限制（如乳糖不耐、无麸质等），请你根据配料表自行判断。
3. 不要编造配料表里没有的成分。
4. 语气温和专业，不要引起不必要的恐慌。

输出格式：
【结论】适合 / 不适合 / 需谨慎
【原因】结合具体配料和用户限制解释
【建议】给用户可执行的建议，如替代品或注意事项
"""
    
    # ===== 第2次大模型调用：DeepSeek API =====
    response = client.chat.completions.create(
        model="deepseek-chat", # 或者用 deepseek-reasoner
        messages=[
            {"role": "system", "content": "你是食品安全与健康助手，请根据配料表和用户限制给出建议。"},
            {"role": "user", "content": prompt}
        ]
    )
    return response.choices[0].message.content

print("主流程定义完成！")
```

```python
def check_compatibility(ingredients, profile):
    """匹配过敏原、疾病风险、饮食偏好，输出风险列表和结论"""
    allergen_synonyms = {
        "花生": ["花生", "花生油", "花生蛋白", "花生酱"],
        "牛奶": ["牛奶", "乳粉", "乳清", "乳蛋白", "奶油", "黄油", "乳糖"],
        "鸡蛋": ["鸡蛋", "蛋清", "蛋黄", "蛋粉"], "大豆": ["大豆", "黄豆", "大豆油"],
        "小麦": ["小麦", "面粉", "麸质", "面筋"], "坚果": ["杏仁", "腰果", "核桃", "坚果"],
        "海鲜": ["虾", "蟹", "鱼", "贝", "海鲜"], "芝麻": ["芝麻"],
    }
    condition_risks = {
        "高血压": ["食盐", "钠", "酱油", "味精", "鸡精"],
        "高血糖": ["糖", "蔗糖", "果葡糖浆", "麦芽糖", "蜂蜜", "糊精"],
        "高血脂": ["油", "奶油", "黄油", "脂肪"], "痛风": ["海鲜", "虾", "蟹", "内脏"],
        "肾病": ["食盐", "钠", "蛋白", "钾", "磷"],
    }
    diet_rules = {
        "低盐": ["食盐", "钠", "酱油", "味精"], "低糖": ["糖", "蔗糖", "蜂蜜"],
        "低脂": ["油", "奶油", "黄油", "脂肪"], "素食": ["肉", "鱼", "虾", "蟹", "鸡", "猪", "牛"],
    }
    
    risks = []
    # 1. 过敏原匹配
    for allergen in profile["allergens"]:
        synonyms = allergen_synonyms.get(allergen, [allergen])
        for ing in ingredients:
            if any(s in ing for s in synonyms):
                risks.append({"type": "过敏原", "target": allergen, "ingredient": ing, "level": "高"})
    # 2. 疾病风险匹配
    for condition in profile["conditions"]:
        risk_words = condition_risks.get(condition, [])
        for ing in ingredients:
            if any(w in ing for w in risk_words):
                risks.append({"type": "疾病风险", "target": condition, "ingredient": ing, "level": "中"})
    # 3. 饮食偏好匹配
    for pref in profile["diet"]:
        risk_words = diet_rules.get(pref, [])
        for ing in ingredients:
            if any(w in ing for w in risk_words):
                risks.append({"type": "偏好冲突", "target": pref, "ingredient": ing, "level": "低"})
    
    # 4. 综合结论
    if any(r["type"] == "过敏原" for r in risks):
        suitable = "不适合"
    elif risks:
        suitable = "需谨慎"
    else:
        suitable = "适合"
    return {"suitable": suitable, "risks": risks}

# 测试函数3
print("\n测试函数3：")
test_profile = parse_user_constraints("我对花生过敏，有高血压")
test_ings = normalize_ingredients("小麦粉、花生、食盐、白砂糖、乳粉")
print(check_compatibility(test_ings, test_profile))
```
